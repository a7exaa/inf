## Что общего у чисел и слов?
Несколько 4-значных чисел в **троичной СС** (алфавит цифр: `0, 1, 2`), с незначащими нулями спереди:

| № по порядку | Число в $3$-й СС | Значение в $10$-й СС | Слово из букв `А, К, Р` |
| :---: | :---: | :---: | :---: |
| **1** | `0000` | $0_{10}$ | `АААА` |
| **2** | `0001` | $1_{10}$ | `АААК` |
| **3** | `0002` | $2_{10}$ | `АААР` |
| **4** | `0010` | $3_{10}$ | `ААКА` |
| **5** | `0011` | $4_{10}$ | `ААКК` |

> **Главный вывод:** Список слов по алфавиту — это обычный счёт в $N$-ичной системе счисления, где:
> * **Основание СС ($N$)** — количество доступных букв в алфавите.
> * **Разрядность числа** — длина слова.
> * **Цифры от $0$ до $N-1$** — буквы в том порядке, в каком они идут в начале списка.

## 🔄 Генерация списка слов с помощью вложенных циклов `for`

Когда в задаче появляются дополнительные условия (например, *«в слове ровно две буквы А и рядом не стоят две гласные»*), считать в системах счисления становится неудобно. Напишем программу, которая сама сгенерирует весь список!



---


```python
# Проверим Задачу 1.1 и 1.2 через генерацию списка вложенными циклами:
alf = 'АПРСУ' # ВАЖНО: буквы строго по алфавиту!
k = 0

for b1 in alf:
    for b2 in alf:
        for b3 in alf:
            for b4 in alf:
                for b5 in alf:
                    word = b1 + b2 + b3 + b4 + b5
                    k += 1 # Увеличиваем номер для КАЖДОГО слова в списке

                    if word == 'ПАРУС':
                        print("Номер слова ПАРУС:", k)
                    if k == 350:
                        print("Слово под номером 350:", word)
```

### 🛠 Шпаргалка по методу `.count()` и проверкам:
* `word.count('А') == 1` — буква `А` встречается **ровно 1 раз**.
* `word.count('А') <= 2` — буква `А` встречается **не более 2 раз** (0, 1 или 2 раза).
* `word.count('А') >= 1` (или `'А' in word`) — буква `А` встречается **хотя бы 1 раз**.
* `'АА' not in word` — две буквы `А` **не стоят рядом**.
* `word[0] != 'Й'` — слово **не начинается** на букву `Й`.
