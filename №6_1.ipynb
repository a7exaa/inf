## 🐢 Шпаргалка по `ColabTurtle`

### Основные команды
* **`initializeTurtle()`** — обязательный запуск холста перед началом работы.
* **`speed=10`** — опараметр задает скорость от 1 до 13.
* **`forward(pixels)`** — движение вперед на указанное количество клеток/пикселей.
* **`backward(pixels)`** — движение назад.
* **`right(angle)`** — поворот вправо на заданный угол в градусах.
* **`left(angle)`** — поворот влево на заданный угол в градусах.
* **`penup()`** — поднимает перо (черепашка движется, не оставляя след).
* **`pendown()`** — опускает перо (черепашка снова рисует линию).

---

### ⚖️ Различия: Google Colab vs Стандартный IDLE (ПК)

| Параметр | Стандартный IDLE (`turtle`) | Google Colab (`ColabTurtle`) |
| :--- | :--- | :--- |
| **Инициализация** | Окно создается автоматически или через `turtle.Screen()`. | Требует обязательного вызова функции `initializeTurtle()`. |
| **Размер холста** | Настраивается через `screen.setup(width, height)`. | Задается при инициализации: `initializeTurtle(initial_width=800, initial_height=500)`. |
| **Скорость** | Задается командой `turtle.speed(n)` (шкала от 0 до 10, где 0 — мгновенная отрисовка). | Задается аргументом при инициализации `initial_speed=n` (шкала от 1 до 10). |
