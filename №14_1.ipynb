## Теория: Перевод чисел уголком → Алгоритм в Python

Чтобы перевести число в любую систему счисления на Python, используется деление нацело (`//`) и взятие остатка (`%`) в цикле `while`:
* `n % base` — текущая младшая цифра.
* `n //= base` — отбрасывание разобранной цифры.
* `while n > 0:` — продолжение цикла, пока число не исчерпано.

```python
# Базовый шаблон перевода и подсчета цифр
n = 29
k = 3
count = 0

while n > 0:
    if n % k == 2:
        count += 1
    n //= k

print("Количество двоек в 3-ичной записи числа 29:", count)
```
