### Задача 1.1. Ветвление по четности $N$ и дописывание срезов (Ловушка: ищем минимальное $R$)
На вход алгоритма подается натуральное число $N \ge 2$. Алгоритм строит по нему новое число $R$:
1. Строится двоичная запись числа $N$.
2. Если число $N$ четное, то к двоичной записи слева (в начало) дописывается `10`, а справа (в конец) — две последние цифры исходной двоичной записи.
3. Если число $N$ нечетное, то к двоичной записи слева дописывается `1`, а справа — `11`.

Полученная таким образом запись является двоичной записью искомого числа R.

4. Результат переводится в десятичную систему и выводится на экран.

Укажите **минимальное** число **$R$**, которое больше $160$ и может являться результатом работы данного алгоритма.


```python
a = []
for n in range(2, 1000):
    b = bin(n)[2:]
    if n % 2 == 0:
        b = '10' + b + b[-2:]
    if n % 2 != 0:
        b = '1' + b + '11'
    r = int(b, 2)
    if r > 160:
        a.append(r)

print(min(a))
```

### Задача 1.2. Классический «Бит четности»
На вход алгоритма подается натуральное число $N$. Алгоритм строит по нему новое число $R$:
1. Строится двоичная запись числа $N$.
2. Складываются все цифры двоичной записи, и остаток от деления суммы на 2 дописывается в конец числа (справа).
3. Над этой записью производятся те же действия — справа дописывается остаток от деления новой суммы цифр на 2.

Полученная таким образом запись является двоичной записью искомого числа R.

4. Результат переводится в десятичную систему и выводится на экран.

Полученная таким образом запись является двоичной записью искомого числа $R$. Укажите **минимальное** число $N$, после обработки которого с помощью этого алгоритма получается число $R > 93$.

```python
for n in range(1, 1000):
    b = bin(n)[2:]
    b += str(b.count('1') % 2)
    b += str(b.count('1') % 2)
    r = int(b, 2)
    if r > 93:
        print(n) # Ответ: 24
        break
```
