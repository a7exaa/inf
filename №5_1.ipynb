## Теория

### Двоичная запись числа

- `bin(n)` возвращает строку с префиксом `0b`.
- `bin(n)[2:]` — двоичная запись без префикса.
- `int(b, 2)` — перевод двоичной строки в десятичное число.

### Индексы и срезы строк

| Запись | Результат для `b = "11010"` |
|---|---|
| `b[0]` | `1` |
| `b[:2]` | `11` |
| `b[-1]` | `0` |
| `b[-2:]` | `10` |
| `b[2:]` | `010` |
| `b[:-2]` | `110` |

```python
# Демонстрация работы со строками и срезами
n = 26
b = bin(n)[2:]
print("Число в двоичной СС:", b)
```

```python
print("Первая цифра b[0]:", b[0])
print("Вторая цифра b[1]:", b[1])
print("Две первые цифры b[:2]:", b[:2])
```

```python
print("Последняя цифра b[-1]:", b[-1])
print("Две последние цифры b[-2:]:", b[-2:])
```

```python
print("Всё, кроме двух первых цифр b[2:]:", b[2:])
print("Всё, кроме двух последних цифр b[:-2]:", b[:-2])
```

### Дописывание и замена отдельных разрядов срезами

Чтобы дописать цифры, соединяем строки через `+`. Чтобы заменить два первых разряда, используем `'10' + b[2:]`; чтобы заменить два последних — `b[:-2] + '01'`.

```python
b = '11010'
print('10' + b[2:])
print(b[:-2] + '01')
```

### Метод `count()`

`b.count('1')` возвращает количество символов `1` в строке `b`. Аналогично `b.count('0')` считает нули. Метод возвращает целое число и не изменяет строку.

Сумма цифр двоичной записи равна количеству единиц. Остаток от деления этой суммы на 2 — `b.count('1') % 2`. Чтобы дописать его к строке, используем `str()`.

```python
b = '11010'
print(b.count('1'))
print(b.count('0'))
print(b.count('1') % 2)
```

```python
b = '11010'
b = b + str(b.count('1') % 2)
print(b)
```

## Списки

Список позволяет хранить несколько значений в одной переменной. В задачах №5 будем сохранять все подходящие значения $N$ или $R$.

- `a = []` — создаём пустой список.
- `a.append(12)` — добавляем число 12 в конец списка.
- `min(a)` — наименьший элемент списка.
- `max(a)` — наибольший элемент списка.

`min()` и `max()` применяем к непустому списку.

```python
a = []
a.append(12)
a.append(5)
a.append(18)
print(a)
print(min(a))
print(max(a))
```

### Заполнение списка в цикле

Сохраним все числа от 1 до 20 включительно, которые делятся на 3.

```python
a = []
for n in range(1, 21):
    if n % 3 == 0:
        a.append(n)

print(a)
print(min(a))
print(max(a))
```

В задаче №5 добавляем `n`, если требуется найти исходное число $N$, и `r`, если требуется найти результат работы алгоритма $R$.

## Разбор задач

### Задача 1

На вход алгоритма подается натуральное число $N \ge 2$. Алгоритм строит по нему новое число $R$:
1. Строится двоичная запись числа $N$.
2. Если число $N$ четное, то к двоичной записи слева (в начало) дописывается `10`, а справа (в конец) — две последние цифры исходной двоичной записи.
3. Если число $N$ нечетное, то к двоичной записи слева дописывается `1`, а справа — `11`.
4. Результат переводится в десятичную систему и выводится на экран.

Укажите **минимальное** число **$R$**, которое больше $160$ и может являться результатом работы данного алгоритма.


```python
a = []
for n in range(2, 1000):
    b = bin(n)[2:]
    if n % 2 == 0:
        b = '10' + b + b[-2:]
    if n % 2 != 0:
        b = '1' + b + '11'
    r = int(b, 2)
    if r > 160:
        a.append(r)

print(min(a))
```

### Задача 2

На вход алгоритма подается натуральное число $N$. Алгоритм строит по нему новое число $R$:
1. Строится двоичная запись числа $N$.
2. Складываются все цифры двоичной записи, и остаток от деления суммы на 2 дописывается в конец числа (справа).
3. Над этой записью производятся те же действия — справа дописывается остаток от деления новой суммы цифр на 2.
4. Результат переводится в десятичную систему и выводится на экран.

Полученная таким образом запись является двоичной записью искомого числа $R$. Укажите **минимальное** число $N$, после обработки которого с помощью этого алгоритма получается число $R > 93$.

```python
a = []
for n in range(1, 1000):
    b = bin(n)[2:]
    b = b + str(b.count('1') % 2)
    b = b + str(b.count('1') % 2)
    r = int(b, 2)
    if r > 93:
        a.append(n)

print(min(a))
```
