### Задача 1 (Попробуйте решить её привычным способом!)
Маша составляет 5-буквенные коды из букв слова **ИГРОК**. При этом **каждая буква используется ровно один раз** (буквы в слове не повторяются), код не может начинаться с буквы **К** и не может содержать сочетания **двух гласных букв**. Сколько различных кодов может составить Маша?

```python
# Попытка через product (Спойлер: буквы повторяются! Приходится городить кучу проверок .count() == 1)
# А теперь решение через новый инструмент — permutations:
from itertools import permutations

k = 0

for p in permutations("ИГРОК", 5):
    word = "".join(p)
    w = word.replace('И', '*').replace('О', '*')

    if word[0] != "К" and "**" not in w:
        k += 1
print("Ответ Задачи 1:", k)
```

### Шпаргалка: `product` против `set(permutations)`

| Критерий | `product("АБВ", repeat=5)` | `set(permutations("АБВГД", 5))` |
| :--- | :--- | :--- |
| **Что делает с буквами?** | Берет буквы **с повторами** (сколько угодно раз) | Перекладывает имеющиеся карточки (**без повторов** позиций) |
| **Слова-маркеры в условии** | *«составляет слова из букв...»*, *«буквы могут повторяться»*, *«список слов по алфавиту»* | *«перестановкой букв слова...»*, *«каждую букву ровно 1 раз»*, *«все цифры различны»* |
| **Как пишем в коде?** | `for p in product("АБВ", repeat=5):` | `for p in set(permutations("СЛОВО", 5)):` |

> **Важное примечание:** Функция `set()` удаляет дубликаты, но перемешивает порядок слов! Поэтому в задачах на **алфавитный список и поиск номера слова** используется `sorted(...)`.
