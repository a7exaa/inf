```python
#задание 5
a = []

for n in range(1, 1000):
  b = bin(n)[2:]

  if n%2 == 0:
    b = '11' + b + '11'
  else:
    b = '1' + b + '00'

  r = int(b, 2)

  if r <= 416:
    a.append(r)

print(max(a))
```

```python
#задание 8
from itertools import product
alf = sorted('ЦИФЕРБЛАТ')

n = 0
last = 0
a = []

for p in product(alf, repeat = 5):
  word = ''.join(p)
  n += 1

  if n%2 != 0 and word[0] not in 'РТЕ' and word.count('И') >= 2:
    a.append(n)

print(max(a))
```
