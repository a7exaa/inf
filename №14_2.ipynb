## Блок 1. Теория: Перевод из $N$-ричной в 10-ричную систему

### 1. Развернутая форма записи числа (Аналитика)
Любое число в системе счисления с основанием $k$ можно перевести в десятичную систему, пронумеровав разряды **справа налево, начиная с нуля**:
$$234_k = 2 \cdot k^2 + 3 \cdot k^1 + 4 \cdot k^0 = 2 \cdot 49 + 21 + 4 = 123_{10}$$

### 2. Функция `int(строка, основание)` в Python
В Python встроена функция `int(str, k)`, которая сама переводит строку `str` из системы с основанием `k` (от 2 до 36) в десятичное число.
* **Главное правило:** Первый аргумент обязательно должен быть **строкой** (`str`), а второй — **числом** (`int`).
* **Алфавит цифр:** В системах с основанием больше 10 вместо цифр $10, 11, 12\dots$ используются латинские буквы:
  `0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ`
  *(A = 10, B = 11, C = 12, D = 13, E = 14, F = 15 ... Z = 35)*.

### 3. Сборка числа с неизвестной цифрой через строки
Если в числе спрятана неизвестная цифра $x$ (например, $123x5_{15}$), мы перебираем $x$ по алфавиту 15-ричной системы и подставляем его внутрь строки через сложение `"123"+x+"5"`:

```python
# Демонстрация работы int()
print("Перевод 234 из 7-ричной в 10-ричную:", int("234", 7))
print("Перевод 1A из 16-ричной в 10-ричную:", int("1A", 16))

# Базовый шаблон для 15-ричной системы счисления (цифры от 0 до E)
for x in "0123456789ABCDE":
    s = "123"+x+"5"
    num = int(s, 15)
    # Выведем первые 3 значения для примера:
    if x in "012":
        print("При x =", x)
        print("Cтрока =", s)
        print("Число в десятичной =", num)
```
