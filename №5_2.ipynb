### 1. Функция перевода в любую СС
Опирается на алгоритм деления «уголком» с циклами `while`, `//` и `%`, который мы подробно разбирали на Уроке 1 по заданию №14.



```python
def binn(n):
    s = ''
    while n > 0:
        s = str(n % 2) + s
        n = n//2
    return s

# Пример проверки работы:
print(binn(13))  # '1101'
print(binn(4))   # '100'
```

### Задача 1. Троичный автомат с умножением остатка
На вход алгоритма подаётся натуральное число $N$. Алгоритм строит по нему новое число $R$ следующим образом:
1. Строится **троичная** запись числа $N$.
2. Далее эта запись обрабатывается по следующему правилу:
   - а) если число $N$ делится на 3, то слева к нему приписывается `1`, а справа — `02`;
   - б) если число $N$ на 3 не делится, то **остаток от деления на 3 умножается на 5**, переводится в троичную запись и дописывается в конец числа.
3. Результат переводится в десятичную систему счисления.

Укажите **минимальное** число $N$, после обработки которого с помощью этого алгоритма получается число $R$, не меньшее $177$.

```python
def tri(n):
    s = ''
    while n > 0:
        s = str(n % 3) + s
        n = n // 3
    return s

a = []
for n in range(1, 1000):
    t = tri(n)

    if n % 3 == 0:
        t = '1' + t + '02'
    else:
        # Остаток от деления умножается на 5 и переводится в 3-ичную СС
        ost = (n % 3) * 5
        t = t + tri(ost)

    r = int(t, 3)

    if r >= 177:
        a.append(n)

print(min(a))
```

### Задача 2. Троичный автомат с суммой цифр и срезами
На вход алгоритма подаётся натуральное число $N$. Алгоритм строит по нему новое число $R$ следующим образом:
1. Строится **троичная** запись числа $N$.
2. Далее эта запись обрабатывается по следующему правилу:
   - а) если число $N$ делится на 3, то к этой записи справа дописываются **две последние троичные цифры**;
   - б) если число $N$ на 3 не делится, то вычисляется **сумма цифр** полученной троичной записи, эта сумма умножается на 2, переводится в троичную систему счисления и дописывается в конец числа.
3. Результат переводится в десятичную систему.

Укажите **минимальное нечётное** число $R$, большее $520$, которое может быть получено с помощью описанного алгоритма.

```python
def tri(n):
    s = ''
    while n > 0:
        s = str(n % 3) + s
        n = n // 3
    return s

a = []
for n in range(1, 1000):
    t = tri(n)

    if n % 3 == 0:
        t = t + t[-2:]
    else:
        sm = sum(int(c) for c in t)
        sm = s.count('1') + s.count('2')*2
        t = t + tri(sm * 2)

    r = int(t, 3)
    # Фильтр: больше 520 И нечётное
    if r > 520 and r % 2 != 0:
        a.append(r)

print(min(a))
```
