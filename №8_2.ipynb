### ЗАДАЧА с прошлого урока
Все 6-буквенные слова из букв Г, Р, А, Н, И, Т записаны в алфавитном порядке и пронумерованы.

**Начало списка:**
1. АААААА
2. АААААГ
3. АААААИ
4. АААААН
5. АААААР
6. АААААТ

Под каким номером в списке идёт ПОСЛЕДНЕЕ слово, которое содержит не менее двух букв И, не содержит букву Г, и его номер в списке кратен 3?


````python
alf = sorted('ГРАНИТ')

n = 0
last = 0

for b1 in alf:
    for b2 in alf:
        for b3 in alf:
            for b4 in alf:
                for b5 in alf:
                    for b6 in alf:
                        word = b1 + b2 + b3 + b4 + b5 + b6
                        n += 1
                        if word.count('И') >= 2 and 'Г' not in word and n % 3 == 0:
                            last = n
print(last)
````

````python
from itertools import product
alf = sorted('ГРАНИТ')

n = 0
last = 0

for p in product(alf, repeat=6):
    word = ''.join(p) # Склеиваем кортеж букв ('А', 'Г', 'И', 'Н', 'Р', 'Т') в строку 'АГИНРТ'
    n += 1
    if word.count('И') >= 2 and word.count('Г') == 0 and n % 3 == 0:
        last = n
print(last)
````

### Приём 2. Задачи на числа в $N$-ичной системе счисления
В задачах про числа («Сколько существует $k$-значных чисел в $N$-ичной СС...») действуют **два золотых правила**:
1. **Число НЕ может начинаться с нуля!** Обязательно пишем первой проверкой: `if word[0] != '0':`.
2. Чтобы проверить, что **никакие две чётные или две нечётные цифры не стоят рядом**, используем тот же приём `.replace()`: заменяем все чётные цифры (`02468ACE`) на `'Ч'`, а нечётные (`13579BDF`) на `'Н'`,

и проверяем `'ЧЧ' not in w and 'НН' not in w`.

### Задача 1 (Базовый `product`)
Сколько существует **6-буквенных** слов, составленных из букв **К, А, Т, Е, Р**, в которых буква **Р** встречается ровно 2 раза, а буква **А** — не более 1 раза?

````python
from itertools import product
alf = sorted('КАТЕР')
n = 0
for p in product(alf, repeat=6):
    word = ''.join(p)
    if word.count('Р') == 2 and word.count('А') <= 1:
        n += 1

print(n)
````

### Задача 0 (permutations)

Мария составляет слова путём перестановки букв слова ПИРАТ.
Сколько существует таких слов, в которых:

* буква Т не стоит на первом месте;
* буквы И и Р не стоят рядом ни в каком порядке (ни ИР, ни РИ)?


````python
from itertools import permutations

alf = 'ПИРАТ'
n = 0

for p in permutations(alf):
    word = ''.join(p)
    if word[0] != 'Т' and 'ИР' not in word and 'РИ' not in word:
        n += 1

print(n)
````

````python
# Посмотрим наглядно, чем отличаются product и permutations:
from itertools import product, permutations
alf = 'КОТ'

print([''.join(p) for p in product(alf, repeat=2)])

print([''.join(p) for p in permutations(alf, 2)])

print([''.join(p) for p in permutations(alf)])
````
